## 0. Why Do Biologists Need R? (Why not Excel?)

Most researchers start analyzing data in Microsoft Excel. But in **Spatial Transcriptomics**—where we measure 20,000 genes across thousands of tissue spots—Excel creates massive roadblocks:

1. **Excel Freezes and Crashes:** Excel is limited to ~1 million rows. Spatial data has tens of millions of data points!
2. **Excel Ruining Gene Names:** Excel automatically changes gene symbols into calendar dates (e.g. `SEPT2` becomes `2-Sep`), permanently corrupting published scientific data.
3. **The "Click" Problem:** If you click buttons in menus to clean data, you cannot remember every click two weeks later.

### Why R is Our Superpower:
* 🚀 **Handles Huge Datasets:** R can process millions of cells and transcripts in seconds.
* 📦 **Built for Biology (Bioconductor):** Biologists around the world built thousands of free tools specifically for spatial data (like `SpatialExperiment` and `Seurat`).
* 📜 **A Reusable Recipe:** Code is a permanent, reproducible recipe. You can analyze a brand-new patient sample in seconds with one click.

## 2. Variables & The Assignment Operator (`<-`)

A **variable** is a named storage container in your computer's memory.

In R, we assign values using `<-` (the "arrow" operator, typed as `<` followed by `-`).


`````r

  variable_name <- value
`````



> 💡 **Why not `=`?** 
> 
> While `=` works in most situations, the R community and Bioconductor coding standards strictly recommend `<-` for variable assignment, reserving `=` for passing arguments inside functions.

#### Example:

`````r
x <- 10
y <- 25

print(x)
print(y)

cat("Total =", x, "+", y, "=", x + y, "\n")
`````

### Rules for Variable Names:
* Must start with a letter (e.g., `gene_count`, `spot_1`).
* Can contain letters, numbers, underscores (`_`), and periods (`.`).
* Cannot start with a number (e.g., `1gene` will cause a syntax error).
* R is **case-sensitive**: `Gene_A`, `gene_A`, and `gene_a` are three completely different variables!

### Hands on!

`````r
project-id <- "V1"
target_gene <- "EPCAM"
1st_sample <- "Visium_S1"

print(project-id)
cat("Target gene:", Target_gene, "| Sample:", 1st_sample, "\n")
`````

#### [Common Mistakes] Invalid variable names and case sensitivity.

`````r
# 1. Hyphens are interpreted as subtraction:
project-id <- "V1"   # Error: object 'project' not found
`````

`````r
# 2. Starting with a number:
1st_sample <- "Visium_S1"  # ERROR: unexpected symbol
`````

`````r
# 3. Case sensitivity trap:
target_gene <- "EPCAM"
print(Target_gene)  # ERROR: object 'Target_gene' not found
`````

#### [Standard] Use snake_case (all lowercase with underscores)

`````r
target_gene <- "EPCAM"
sample_01_id <- "Visium_S1"

cat("Target gene:", target_gene, "| Sample:", sample_01_id, "\n")
`````

## 3. The 4 Core Atomic Data Types

Every piece of data in R belongs to a specific type. 

In bioinformatics, you will encounter 4 main atomic types:

| Data Type | Description | Biological Example |
| :--- | :--- | :--- |
| **`numeric`** | Decimal / floating-point numbers (default in R) | `mito_percent <- 8.5` |
| **`integer`** | Whole numbers without decimals (suffixed with `L`) | `total_spots <- 4992L` |
| **`character`** | Text strings enclosed in double quotes `"` | `gene_symbol <- "GAPDH"` |
| **`logical`** | Boolean truth values: `TRUE` or `FALSE` (no quotes!) | `is_expressed <- TRUE` |

You can check the type of any variable using `class()` or `typeof()`.

`````r
mito_percent <- 8.5     # numeric
total_spots  <- 4992L   # integer
gene_symbol  <- "GAPDH" # character
is_expressed <- TRUE    # logical
`````

`````r
typeof(mito_percent)
`````

`````r
# [Common Gotcha] Implicit Type Coercion
# In R, a simple vector c(...) can only hold ONE data type.
# If you mix numbers and strings, R silently converts everything to character!

coerced_data <- c("Gene_A", 150, TRUE)
print(coerced_data)

# Notice how 150 became "150" and TRUE became "TRUE"!
cat("Class of coerced_data:", class(coerced_data), "\n")
`````

## 4. Arithmetic & Biological Calculations

R comes equipped with standard mathematical operators and built-in functions:

| Operator / Function | What It Does | Biological Code Example |
| :--- | :--- | :--- |
| `+` | Addition | `total_reads <- 12000 + 3500` |
| `-` | Subtraction | `bg_spots <- 4992 - 3820` |
| `*` | Multiplication | `mito_pct <- 0.074 * 100` |
| `/` | Division | `depth_scale <- 45 / 15000` |
| `^` | Exponentiation (power) | `fc <- 2^3` |
| `%%` | Modulo (remainder) | `batch_remainder <- 10 %% 3` |
| `%/%` | Integer division | `cohort_batches <- 10 %/% 3` |
| `sqrt()` | Square root | `spot_distance <- sqrt(144)` |
| `log2()` | Base-2 logarithm | `log2_fc <- log2(8)` |
| `round()` | Round decimals | `p_val <- round(0.003412, 3)` |


### Library Size Normalization & Log Transformation
In transcriptomics, raw counts must be normalized by the cell's total sequencing depth.

`````r
raw_gene_count <- 32
total_spot_umi <- 12500
`````

`````r
# Step 1: Calculate Counts Per Million (CPM)
cpm <- (raw_gene_count / total_spot_umi) * 1e6 # -> 1,000,000
`````

`````r
# Step 2: Log2 Transformation (with pseudo-count +1 to prevent log(0))
log2_cpm <- log2(cpm + 1)
`````

`````r

cat("Raw UMI Count:  ", raw_gene_count, "\n")
cat("Total Spot UMIs:", total_spot_umi, "\n")
cat("Normalized CPM: ", round(cpm, 2), "\n")
cat("Log2(CPM + 1):  ", round(log2_cpm, 3), "\n")
`````

#### [Common Gotcha] Division by zero and log(0)
Dividing by zero does not crash R; it returns `Inf` (Infinity):


`````r
10 / 0
`````

Taking the logarithm of 0 gives `-Inf` (Negative Infinity):


`````r
log(0)
`````

#### 💡 The Fix: Pseudo-counts (+1) in Bioinformatics
If an unexpressed gene has 0 reads, `log2(0)` is `-Inf` (which breaks statistical models).
Adding `+1` ensures that 0 counts become exactly 0 after transformation:


`````r
log2(0 + 1)
`````

## 5. Working Directory & Path Management

A **working directory** is the default folder where R looks for input files (like CSV tables) and saves output files (like PDF plots).

`getwd()`: **Get** **W**orking **D**irectory (shows your current location).

`````r
getwd()
`````

`dir()` or `list.files()`: Lists all files in the current directory.

`````r
dir()
`````

`dir.create("folder_name")`: Creates a new **folder** (e.g. a `_temp` folder for scratch files).


`````r
dir.create("_temp")
`````

`dir.exists("folder_name")`: Confirms whether the `_temp` folder was created.


`````r
dir.exists("_temp")
`````

`unlink("folder_name", recursive = TRUE)`: Deletes a **folder** and all contents inside it when done.


`````r
unlink("_temp", recursive = TRUE)
dir.exists("_temp") # Should return FALSE.
`````

#### 💡 Storing Function Results & Inspecting with `head()`

1. **Storing results:** You can store the output of any function into a variable (e.g. `local_files <- list.files()`) so you can reuse, count, or filter it later.
2. **`head(x, n = 5)`:** In bioinformatics, directories or data tables can contain thousands of items. Instead of printing everything and flooding your screen, `head()` displays only the first few items.

`````r
# 1. Store the function's output in a variable:
local_files <- list.files()
`````

`````r
# 2. Check total count using length():
cat("Detected", length(local_files), "items in folder.\n")
`````

`````r
# 3. Preview just the first 5 items using head():
head(local_files, 5)
`````

#### [Common Mistakes] File Path Separators: Windows vs Linux/Mac

On Windows, copied file paths often use backslashes: `"folder\hello.txt"`.  
In R, a single backslash `\` is an **escape character** (like `\n` for newline). Using a single `\` with normal text causes an immediate error:  
`Error: '\h' is an unrecognized escape in character string`

> 💡 **Rule of Thumb:** **ALWAYS** use forward slashes `"/"` in R strings, even on Windows!

`````r
# ❌ TRAP: Backslash causes an R escaping error (uncomment to test):
# invalid_path <- "folder\hello.txt"  # Error: '\h' is an unrecognized escape!

incorrect_path <- ".\look_here\hello.txt"

cat("Path:        ", incorrect_path, "\n")
cat("File exists? ", file.exists(safe_path), "\n")
`````

`````r
# ✅ SAFE: Forward slash works seamlessly across Windows, Mac, and Linux:
safe_path <- "./look_here/hello.txt"

cat("Path:        ", safe_path, "\n")
cat("File exists? ", file.exists(safe_path), "\n")
`````
